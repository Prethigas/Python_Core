```python
#Property Getter
class Student:
    def __init__(self, total):
        self.value = total

    @property
    def total(self):
        return self.value
stud01 = Student(450)
print(stud01.total)       
```

```python
#Property Setter
class Student:
    def __init__(self, total):
        self.value = total

    @property
    def total(self):
        return self.value

    @total.setter
    def total(self, t):
        print('Setter Function Called Successfully!....')
stud01 = Student(450)
print(stud01.total)
stud01.total = 20  
```

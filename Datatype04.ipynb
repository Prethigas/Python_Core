```python
v1 = {1, 2, 3, 4, 5, 1, 2}
v2 = {8, 4, 9, 5, 6, 7, 7}
v1.update(v2)
print(v1)
```

```python
v1 = {1, 2, 3, 4, 5, 1, 2, 8}
v2 = {4, 5, 6, 7, 9, 4, 1}
# Union
uV1 = v1.union(v2)
print(v1.union(v2))
print(uV1)
```

```python
v1 = {1, 2, 3, 4, 5, 1, 2, 8}
v2 = {4, 5, 6, 7, 9, 4, 1}
# Intersection
uV1 = v1.intersection(v2)
print(uV1)
```

```python
v1 = {1, 2, 3, 4, 5, 1, 2, 8}
v2 = {4, 5, 6, 7, 9, 4, 1}
# Symmetric_Difference
uV1 = v1.symmetric_difference(v2)
print(uV1)
```

```python
a={1,2,3}
b={1,4,5}
print(type(a))
print(a.union(b))
print(a|b)
print(a.intersection(b))
print(a&b)
print(a.difference(b))
print(a-b)
print(a.symmetric_difference(b))
print(a^b)
```

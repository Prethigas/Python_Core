```python
x = 10  
def modify_global():
    global x 
    x = x + 5
    print("Inside function:", x)
modify_global()
print("Outside function:", x)
```

```python
x = 10  
def local_example():
    nonlocal x 
    print("Inside function:", x)
local_example()
print("Outside function:", x)
```

```python
def outer():
    y = 10  # Enclosing function variable
    def inner():
        nonlocal y
        y += 5
        print("Inside inner:", y)
    inner()
    print("Inside outer:", y)
outer()
```

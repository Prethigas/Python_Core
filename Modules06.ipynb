```python
print(math.sin(math.radians(30)))   # 0.4999999999999999
print(math.cos(math.radians(60)))   # 0.5000000000000001
print(math.degrees(math.pi))        # 180.0
print(math.radians(180))            # 3.141592653589793
```

```python
print(math.fsum([0.1, 0.2, 0.3]))   # 0.6
print(math.prod([1, 2, 3, 4]))      # 24
print(math.hypot(3, 4))             # 5.0
print(math.copysign(3, -1))         # -3.0
print(math.dist([0, 0], [3, 4]))    # 5.0
```

```python
print(math.isnan(float('nan')))    # True
print(math.isinf(float('inf')))    # True
print(math.isfinite(100))          # True
print(math.isclose(1.0000001, 1.0))  # False (unless tolerance given)
print(math.isclose(1.0000001, 1.0, rel_tol=1e-3))  # True
```

```python
print(math.pi)     # 3.141592653589793
print(math.e)       # 2.718281828459045
print(math.tau)     # 6.283185307179586
print(math.inf)     # inf
print(math.nan)     # nan
```

```python
import requests

payload = {'username': 'arun', 'password': '12345'}

response = requests.post('https://httpbin.org/post', json=payload)

print("Status Code:", response.status_code)
print("Response Text:", repr(response.text))   
```

```python
headers = {'Authorization': 'Bearer YOUR_TOKEN'}

response = requests.get('https://api.github.com/user', headers=headers)

print(response.status_code)
```

```python
response = requests.get('https://api.github.com/invalid-endpoint')

if response.status_code == 200:
    print("Success")
else:
    print("Failed:", response.status_code)

# Alternative: raises an exception automatically for 4xx/5xx errors
try:
    response.raise_for_status()
except requests.exceptions.HTTPError as e:
    print("Error:", e)
```

```python
# PUT - update
response = requests.put('https://httpbin.org/put', data={'name': 'Updated'})
print(response.status_code)

# DELETE - remove
response = requests.delete('https://httpbin.org/delete')
print(response.status_code)
```

```python
try:
    response = requests.get('https://api.github.com', timeout=5)
    print(response.status_code)
except requests.exceptions.Timeout:
    print("Request timed out")
except requests.exceptions.ConnectionError:
    print("Connection error")
```

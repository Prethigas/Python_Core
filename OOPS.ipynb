```python
#Class Declaration
class Student:
    name = 'Sara' # or Student.name = 'Lara'
    age = 25
    
# Dot Method
print(Student.name)
del(Student.name)
# get Attribute Method
print(getattr(Student, 'age'))
delattr(Student, 'age')
print(getattr(Student, 'gender', 'male'))

# Set Attribute Method
setattr(Student, 'gender', 'female')
print(Student.gender)
```

```python
The __init__() method is used to assign values to object properties, or to perform operations that are necessary when the object is being created.

The self parameter is a reference to the current instance of the class.
It is used to access properties and methods that belong to the class.
It is First parameter of any method in the class.
It links the method to the specific object

self.name = name --> assign a value to an instance variable inside a class.
    |
Creates or updates an instance attribute called name for that object.

class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

p1 = Person("Emil", 36)

print(p1.name)
print(p1.age)
```

```python
#A method that accesses object properties:

class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def get_info(self):
    return f"{self.name} is {self.age} years old"

p1 = Person("Tobias", 28)
print(p1.get_info())
```

```python
# With the __str__() method:

class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def __str__(self):
    return f"{self.name} ({self.age})"

p1 = Person("Tobias", 36)
print(p1)
```

```python
# Single Inheritance
# Parent class
class Person:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname

    def print_name(self):
        print(f"{self.firstname} {self.lastname}")

# Child class inheriting from Person
class Student(Person):
    def __init__(self, fname, lname, year):
        # Call the parent class constructor
        super().__init__(fname, lname)
        self.graduationyear = year

    def print_student_info(self):
        print(f"{self.firstname} {self.lastname}, Graduation Year: {self.graduationyear}")

# Create a Student object
x = Student("Mike", "Olsen", 2019)

# Display information
x.print_name()            # From Person class
x.print_student_info()    # From Student class
```

```python
# Multiple Inheritance
class Father:
    def playing(self):
        print('Father have Playing Chess!...')

    def ownBike(self):
        print('Father have Own Bike!...')

class Mother:
    def playing(self):
        print('Mother have Playing Chess!...')

    def cooking(self):
        print('Mother have Cooking Skill!...')

class Son(Father, Mother):
    def reading(self):
        print('Son have Reading Skill!...')

# Create the Object
object01 = Son()
object01.reading()
object01.playing()
object01.cooking()
object01.ownBike()
```

```python
# Multi Level Inheritance
class Father:
    def playing(self):
        print('Father have Playing Chess!...')

    def ownBike(self):
        print('Father have Own Bike!...')

class Mother(Father):
    def playing(self):
        print('Mother have Playing Chess!...')

    def cooking(self):
        print('Mother have Cooking Skill!...')

class Son(Mother):
    def reading(self):
        print('Son have Reading Skill!...')

# Create the Object
object02 = Son()
object02.reading()
object02.playing()
object02.cooking()
object02.ownBike()
```

```python
#Hierarchical inheritance:
class Parent:
    def display(self):
        print("Parent")

class Child(Parent):
    def child_display(self):
        print("Child")

c = Child()
c.display()
c.child_display()

class Son(Parent):
    def son_display(self):
        print("son")

s = Son()
s.son_display()
s.display()
```

```python
#Diamond Problem:
class A:
    def display(self):
        print('I am a Class Of A')

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass

obj01 = D()
obj01.display()
```

```python
x = 10  # Global variable
def local_example():
    x = 5  # Local variable (shadows global x)
    x = x+5
    print("Inside function:", x)
local_example()
print("Outside function:", x)
```

```python
x = 10  
def modify_global():
    global x  # Declare that we are using the global variable
    x = x + 5
    print("Inside function:", x)
modify_global()
print("Outside function:", x)
```

```python
def outer():
    y = 10  # Enclosing function variable
    def inner():
        nonlocal y
        y += 5
        print("Inside inner:", y)
    inner()
    print("Inside outer:", y)
outer()
```

```python
# Overwrite (or create) a file
with open("example.txt", "w") as file:
    file.write("Hello, World!\n")
    file.write("This is a new line.")

# Append to an existing file
with open("example.txt", "a") as file:
    file.write("\nAppended text.")
```

```python
file = open("example.txt", "r")
print("File opened successfully")
file.close()
```

```python
# Read entire file
with open("example.txt", "r") as file:
    content = file.read()
    print(content)

# Read line by line
with open("example.txt", "r") as file:
    for line in file:
        print(line.strip())

# Read all lines into a list
with open("example.txt", "r") as file:
    lines = file.readlines()
    print(lines)
    print(lines[2])
```

```python
import csv

# Writing CSV
with open("data.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Name", "Age"])
    writer.writerow(["Alice", 30])

# Reading CSV
with open("data.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
#Without .strip():
with open("example.txt", "r") as f:
    para = f.readlines()

print(para)

for i in para:
    print(i)
```

```python
#With .strip(): removes the \n (and any leading/trailing spaces) before print() adds its own.
with open("para.txt", "r") as f:
    para = f.readlines()

for i in para:
    print(i.strip())
```

```python
#repr() here is useful because it shows the \n explicitly instead of rendering it as an actual line break
line = "Python is easy\n"
print(repr(line))          # 'Python is easy\n'
print(repr(line.strip()))  # 'Python is easy'
```

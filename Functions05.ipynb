```python
# Definition
def greet():
    print("Hello!")

# Calling
greet()
```

```python
def add(a, b):     # a, b => parameters
    return a + b

add(5, 10)          # 5, 10 => arguments
```

```python
def add_all(*numbers):
    total = 0
    for n in numbers:
        total += n
    return total

print(add_all(1, 2, 3, 4))
```

```python
def student_info(name, age):
    print(f"Name: {name}, Age: {age}")

student_info(age=21, name="Priya")
```

```python
def print_details(**info):
    for key, value in info.items():
        print(f"{key}: {value}")

print_details(name="Ravi", age=25, city="Chennai")
```

```python
def greet(name="Guest"):
    print(f"Hello, {name}!")

greet()
greet("Kumar")
```

```python
def display_fruits(fruits):
    for f in fruits:
        print(f)

fruit_list = ["Apple", "Banana", "Mango"]
display_fruits(fruit_list)

# Unpacking a list into *args
def add_all(*numbers):
    return sum(numbers)

nums = [1, 2, 3, 4]
print(add_all(*nums))
```

```python
square = lambda x: x * x
print(square(5))

add = lambda a, b: a + b
print(add(3, 4))
```

```python
currentDate = dt.date.today()
print('Current Date : ', currentDate)

userDate = dt.date(1999, 7, 18)
print('User Date : ', userDate)

print('-----------------------------------')

DiffTwoDate = currentDate - userDate
print(DiffTwoDate)

print(DiffTwoDate.days) #pull out just the number of days
```

```python
currentDateTime = dt.datetime.now()
print(currentDateTime)
print('----------------------------------')

format01 = currentDateTime.strftime('%A')       # Full weekday name
print(format01)

format02 = currentDateTime.strftime('%B')       # Full month name
print(format02)

format03 = currentDateTime.strftime('%d/%b/%Y') # day/Mon/Year
print(format03)

format04 = currentDateTime.strftime('%D - %B : %Y')
print(format04)
```

```python
dateString = "18-07-1999"

convertedDate = dt.datetime.strptime(dateString, '%d-%m-%Y')

print(convertedDate)
print(type(convertedDate))
```

```python
currentDate = dt.date.today()
print('Today : ', currentDate)

futureDate = currentDate + dt.timedelta(days=10)
print('After 10 days : ', futureDate)

pastDate = currentDate - dt.timedelta(days=30)
print('30 days ago : ', pastDate)
```
